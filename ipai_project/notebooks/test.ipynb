```python
import h5py
import os
import numpy as np

# 1. Настройка путей
folder = r'C:\MariaSamosudova\Projects\UNIVER\repos\ELIQSIR\data\datasets'
file_name = 'advanced_quantum_dataset.h5'
full_path = os.path.join(folder, file_name)

print(f"📡 Подключение к базе: {file_name}")

try:
    with h5py.File(full_path, 'r') as f:
        keys = list(f.keys())
        total = len(keys)
        
        # 2. Выбираем молекулу для анализа
        # Можешь поменять индекс [-1] на любой другой, например [50000]
        target_id = keys[-1] 
        mol = f[target_id]
        
        print(f"📊 Статистика: {total} молекул в файле.")
        print(f"🧬 Детальный разбор молекулы ID: {target_id}")
        print("-" * 50)

        # 3. Автоматический обход всех доступных данных
        available_fields = list(mol.keys())
        
        for field in ['x', 'pos', 'edge_index', 'edge_attr', 'u']:
            if field in mol:
                data = mol[field][:]
                
                print(f"➡️ Поле [{field.upper()}]:")
                print(f"   • Форма (shape): {data.shape}")
                print(f"   • Тип данных: {data.dtype}")
                
                # Выводим кусочек данных для визуальной проверки
                if field == 'x':
                    print(f"   • Атомы (первые 2): \n{data[:2]}")
                elif field == 'pos':
                    print(f"   • 3D Координаты (первые 2): \n{data[:2]}")
                elif field == 'u':
                    print(f"   • Глобальные фичи [LogP, TPSA, RotBonds]: {data[0]}")
                elif field == 'edge_index':
                    print(f"   • Связи (фрагмент): {data[:, :5]}")
            else:
                print(f"⚠️ Поле [{field.upper()}]: ОТСУТСТВУЕТ (вероятно, старая запись)")

        print("-" * 50)
        print(f"✅ Инспекция завершена. Файл здоров.")

except Exception as e:
    print(f"❌ Ошибка доступа: {e}")
    print("Совет: Убедись, что путь верный и буква 'r' перед строкой пути на месте.")
```

```python
import h5py
import numpy as np
import os
import random

# 1. Path to your production dataset
h5_path = r'C:\MariaSamosudova\Projects\UNIVER\repos\ELIQSIR\data\datasets\advanced_quantum_dataset.h5'

print(f"🕵️‍♂️ Starting Deep Audit of: {h5_path}")

if not os.path.exists(h5_path):
    print("❌ File not found! Check the path or wait for the first batch to be saved.")
else:
    with h5py.File(h5_path, 'r') as f:
        keys = list(f.keys())
        total = len(keys)
        print(f"📊 Total molecules in file: {total}")
        
        if total > 0:
            # Pick a RANDOM molecule to see different data on each run
            sample_id = random.choice(keys)
            mol = f[sample_id]
            print(f"🧬 Inspecting Random Molecule ID: {sample_id}")
            print("-" * 50)

            # --- FIELD 1: Node Features (X) ---
            # Contains atomic numbers, charges, hybridization, etc.
            x = mol['x'][:]
            print(f"➡️ [X] Node Features: Shape {x.shape} (Atoms, Features)")
            if x.shape[1] == 10:
                print("   ✅ Feature count is correct (10/10)")
                print(f"   • Gasteiger Charges sample: {x[:3, 6]}")

            # --- FIELD 2: 3D Positions (POS) ---
            pos = mol['pos'][:]
            print(f"➡️ [POS] 3D Coordinates: Shape {pos.shape}")
            if not np.all(pos == 0):
                print("   ✅ 3D Geometry is present")

            # --- FIELD 3: Adjacency (EDGE_INDEX) ---
            ei = mol['edge_index'][:]
            print(f"➡️ [EDGE_INDEX]: Shape {ei.shape} (2, Total_Bonds)")

            # --- FIELD 4: Global Descriptors (U) ---
            # [LogP, TPSA, RotatableBonds, ExactMW]
            u = mol['u'][:]
            print(f"➡️ [U] Global Context: Shape {u.shape}")
            print(f"   • Values [LogP, TPSA, RotBonds, MW]: {u[0]}")

            # --- FIELD 5: SMILES String ---
            smiles = mol['smiles'][()]
            if isinstance(smiles, bytes):
                smiles = smiles.decode('utf-8')
            print(f"➡️ [SMILES]: {smiles}")

            print("-" * 50)
            print("💎 VERIFICATION COMPLETE: Data integrity is solid.")
        else:
            print("⚠️ File is empty yet. Keep the pipeline running!")
```
